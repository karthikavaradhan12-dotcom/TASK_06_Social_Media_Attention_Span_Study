```python
Q1 = df['Short_Video_Engagement'].quantile(0.25)

Q3 = df['Short_Video_Engagement'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR

upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['Short_Video_Engagement'] < lower_bound) |
    (df['Short_Video_Engagement'] > upper_bound)
]

outliers
```

```python
z_scores = np.abs(
    stats.zscore(df['Short_Video_Engagement'])
)

outliers_z = df[z_scores > 3]

outliers_z
```

```python
df['Content_Type'].value_counts()
```

```python
from scipy.stats import f_oneway

groups = []

for content in df['Content_Type'].unique():

    group = df[
        df['Content_Type'] == content
    ]['Short_Video_Engagement'].dropna()

    groups.append(group)

f_stat, p_value = f_oneway(*groups)

print("F-Statistic:", f_stat)
print("P-Value:", p_value)
```

```python
if p_value < 0.05:
    print("Reject Null Hypothesis")
    print("Content type significantly affects engagement")
else:
    print("Fail to Reject Null Hypothesis")
    print("Content type does not significantly affect engagement")
```

# Business Insights

## 1. Short-Form Video Content Generates Higher Engagement

The analysis shows that users spend more time engaging with short-form video content such as reels, shorts, and entertainment videos. These formats capture attention quickly and encourage continuous scrolling behavior.

---

## 2. Excessive Social Media Usage Reduces Attention Span

Users spending more hours daily on social media reported lower concentration levels and frequent distraction. Heavy social media usage negatively impacts focus and productivity.

---

## 3. Notifications Strongly Influence User Behavior

Notifications play a major role in bringing users back to social media platforms repeatedly throughout the day. This increases engagement but also interrupts work and study activities.

---

## 4. Continuous Scrolling Behavior is Highly Common

Many users admitted to scrolling continuously without a specific purpose. Infinite scrolling features contribute to addictive usage behavior and increased screen time.

---

## 5. Entertainment Content is Preferred Over Educational Content

Most users showed greater engagement toward entertainment-based content compared to educational or informational content. Quick entertainment creates faster user satisfaction.

---

## 6. Heavy Users Experience Digital Fatigue

Users with higher screen time reported:

* mental tiredness
* reduced productivity
* poor sleep quality
* lower attention span

This indicates negative effects of excessive social media consumption.

---

## 7. Social Media Usage Affects Sleep Quality

Late-night scrolling and prolonged screen exposure negatively affect user sleep patterns and overall well-being.

---

## 8. Content Type Significantly Affects User Engagement

Statistical analysis indicates that different content types produce different engagement levels. Interactive and short-form content retains users for longer durations.

---

## Overall Conclusion

The study demonstrates that social media platforms strongly influence user attention behavior. Features such as short videos, notifications, and endless scrolling increase engagement but may negatively affect attention span, productivity, and digital well-being.
